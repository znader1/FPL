```python
import os
import pandas as pd
pd.options.display.max_columns=64
```

```python
directory  = '../data/Files/Players/'
```

```python
os.listdir(directory)
```

```python
#players_data_fpl_23_24 = pd.DataFrame()
appended_data = []
for f in os.listdir(directory):
    if 'csv' in f:
        data = pd.read_csv(os.path.join(directory, f))
        data['date'] = f[:10]
        appended_data.append(data)
        
players_data_fpl_23_24 = pd.concat(appended_data)
    
```

```python
players_data_fpl_23_24['date'].value_counts(dropna=False).sort_index()
```

```python
players_data_fpl_23_24 = players_data_fpl_23_24[players_data_fpl_23_24['date'] > '2023-08-01']
```

```python
players_data_fpl_23_24[players_data_fpl_23_24['second_name'] == 'Haaland']
```
