## IV. Merge Data in FPL

```python
FPL_diff_seasons = pd.merge(FPL_data, FPL_data_1920, how='left',
                            on=['code', 'web_name'], suffixes=('_20_21','_19_20'))
```

```python
# Merge the two tables in Dataframes
FPL_team_2021 = pd.merge(FPL_data, teams_data, how='left',
                     left_on='team_code', right_on='code')
```

```python
FPL_diff_seasons['cost_diff'] = FPL_diff_seasons['now_cost_20_21'] - FPL_diff_seasons['now_cost_19_20']
```

```python
def plot_cost_diff_top_n(n):
    fig = plt.figure()
# Divide the figure into a 2x2 grid, and give me the first section
    ax1 = fig.add_subplot(221)
    # Divide the figure into a 2x2 grid, and give me the second section
    ax2 = fig.add_subplot(222)
    # Divide the figure into a 2x2 grid, and give me the third section
    ax3 = fig.add_subplot(223)
    # Divide the figure into a 2x2 grid, and give me the third section
    #ax4 = fig.add_subplot(224)

    data = FPL_diff_seasons\
    [['now_cost_20_21', 'now_cost_19_20', 'cost_diff', 'web_name']]\
    .set_index('web_name')
    
    #fig, (ax1, ax2, ax3, ax4) = plt.subplots(2, 2)
    data.sort_values(by='now_cost_19_20', ascending=False)[:n].iloc[::-1].plot.barh(y='now_cost_19_20',ax=ax1,figsize=(20,20), color='red');
    for i, v in enumerate(data.sort_values(by='now_cost_19_20',ascending=False)[:n]['now_cost_19_20'].iloc[::-1]):
        ax1.text(v +0.1, i , str(v), color='black', fontweight='bold')
    data.sort_values(by='now_cost_20_21', ascending=False)[:n].iloc[::-1].plot.barh(y='now_cost_20_21',ax=ax2, color='green');
    for i, v in enumerate(data.sort_values(by='now_cost_20_21',ascending=False)[:n]['now_cost_20_21'].iloc[::-1]):
        ax2.text(v + 0.1, i , str(v), color='black', fontweight='bold')
    data.sort_values(by='cost_diff', ascending=False)[:n].iloc[::-1].plot.barh(y='cost_diff', ax=ax3, color='blue');
    for i, v in enumerate(data.sort_values(by='cost_diff',ascending=False)[:n]['cost_diff'].iloc[::-1]):
        ax3.text(v + 0.1, i , str(v), color='black', fontweight='bold')
    #data.sort_values(by='now_cost', ascending=False)[:n].plot.bar(y='now_cost', ax=ax4, color='orange');
#fig, (ax1, ax2, ax3, ax4) = plt.subplots(2, 2)
```

```python
plot_cost_diff_top_n(20)
```

```python
FPL_team_2021['roi'] = FPL_team_2021['points_per_game']/FPL_team_2021['now_cost']
FPL_team_2021['roi_total'] = FPL_team_2021['total_points']/FPL_team_2021['now_cost']
FPL_team_2021['roi_total_minutes_100'] = FPL_team_2021[FPL_team_2021['total_points']> 100]['total_points']/FPL_team_2021['now_cost']/FPL_team_2021['minutes']
FPL_team_2021['roi_total_minutes_150'] = FPL_team_2021[FPL_team_2021['total_points']> 150]['total_points']/FPL_team_2021['now_cost']/FPL_team_2021['minutes']
FPL_team_2021['roi_total_minutes_120'] = FPL_team_2021[FPL_team_2021['total_points']> 120]['total_points']/FPL_team_2021['now_cost']/FPL_team_2021['minutes']
```

```python
plot_top_n_roi(20)
```

III. Per minute

```python
FPL_team_2021['roi_total_minutes_2200'] = FPL_team_2021[FPL_team_2021['minutes']> 2200]['total_points']/FPL_team_2021['now_cost']/FPL_team_2021['minutes']
```

```python
FPL_team_2021.sort_values(by='roi_total_minutes_2200', ascending=False)[['web_name', 'total_points', 'roi_total', 'minutes', 'ep_next']].head(25)
```

```python
selected_features = ['web_name', 'first_name', 'second_name', 'name','short_name', 'element_type',
                     'ep_next', 'minutes', 'status', 'now_cost', 'roi_total', 'roi_total_minutes_100','points_per_game',
                     'total_points','roi_total_minutes_2200',
                     'roi_total_minutes_150', 'roi_total_minutes_120','strength_overall_home',
                    'strength_overall_away', 'strength_attack_home', 'strength_attack_away',
                    'strength_defence_home', 'strength_defence_away']
```

```python
FPL_final_minutes = FPL_team_2021[selected_features].sort_values(by='roi_total_minutes_2200', ascending=False)
```

```python
FPL_final = FPL_final_minutes.rename(columns={"web_name": "player_name", "name":"team_name", 
                                      'element_type':'player_position', 'now_cost':'price'})
```
