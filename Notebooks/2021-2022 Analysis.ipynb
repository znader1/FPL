```python
def import_csv_players(directory='/Users/ziadNader/OneDrive - Capgemini/Personal Projects/Fantasy Premier League/Data/Files/Players',first_date=time_start, last_date=today_date):
    """function that will import all the csv of players"""
    final_data = pd.DataFrame()
    for filename in os.listdir(directory):
        f = os.path.join(directory, filename)
        # checking if it is a file
        if (os.path.isfile(f)) & ('players' in f) & (os.path.getctime(f) > time_start):
            #print(f)
            #print(time.ctime(os.path.getctime(f)))
            df = pd.read_csv(f)
            final_data = final_data.append(df,ignore_index=True)
            final_data = final_data.sort_values(by='date').reset_index(drop=True)
    return final_data

    
    
    
    
    
```
